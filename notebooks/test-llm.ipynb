```python
import os
from openai import AzureOpenAI
from shared_functions import find_repo_root, read_secret_file
```

```python
REPO_ROOT = find_repo_root()
AZURE_API_ENDPOINT_PATH = REPO_ROOT / "apis" / "secrets" / "azure_api_endpoint.txt"
AZURE_API_KEY_PATH = REPO_ROOT / "apis" / "secrets" / "azure_api_key.txt"
```

```python
subscription_key = read_secret_file(AZURE_API_KEY_PATH)
endpoint = read_secret_file(AZURE_API_ENDPOINT_PATH)
api_version = "2024-12-01-preview"
```

```python
# If no temperature is allowed by Azure, the value is 1.0, else 0.1
{
    "gpt-5-mini": {"allowed_temp": 1.0},
    "gpt-5.1-chat": {"allowed_temp": 1.0},
    "gpt-5.3-chat": {"allowed_temp": 1.0},
    "grok-4-20-non-reasoning": {"allowed_temp": 0.1},
    "grok-4-fast-reasoning": {"allowed_temp": 0.1},
}
```

```python
model_name = "gpt-5.1-chat"
```

```python
client = AzureOpenAI(
    api_version=api_version,
    azure_endpoint=endpoint,
    api_key=subscription_key,
)

response = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": "This is a test of the Azure OpenAI client in Python.",
        },
        {
            "role": "user",
            "content": "Hello!",
        }
    ],
    max_completion_tokens=500,
    temperature=0.7,
    model=model_name
)

print(response.choices[0].message.content)
```
